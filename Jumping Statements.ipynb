```python
# 1. Print numbers from 1 to 10 except 3 and 8

for i in range(1,11):
    if i == 3 or i == 8:
        continue
    print (i)
```

```python
# 2. Find the first number divisible by 7 between 1 and 100

for i in range(1,101):
    if i % 7 == 0:
        print(i)
        break
```

```python
# 3. Count vowels in a string

s = "mayankpatelmayurpatel"
vowel = 0
vowels = "aeiou"
for i in s:
    if i in vowels:
        vowel+=1

print (vowel)
```

```python
# 4. Print all numbers from 1 to 30 except multiples of 5

for i in range(1,31):
    if i % 5 == 0:
        continue
    print (i)
```

```python
# 5. Print numbers 1–50 but skip multiples of 3, skip numbers ending with 7, stop completely if number becomes 40

for i in range(1,51):
    if i % 3 == 0 or i % 10 == 7:
        continue
    if i == 40:
        break
    print (i)
```
