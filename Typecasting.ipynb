## Type casting, also known as type conversion, is the process of converting a value from one data type to another in Python.

1. int --> float, string, boolean, complex
   
- int --> float (Output: float value)
- int --> string (Output: string value)
- int --> boolean (Output: boolean value)
- int --> complex (Output: complex value)

2. float --> int, string, boolean, complex

- float --> int (Output: int float)
- float --> string (Output: string value)
- float --> boolean (Output: boolean value)
- float --> complex (Output: complex value)

3. string(if string = "123") --> int, float, boolean, complex
   
- string --> int (Output: int value)
- string --> float (Output: float value)
- string --> boolean (Output: boolean value)
- string --> complex (Output: complex value)

4. string(if string = "abc") --> int, float, boolean, complex
   
- string --> int (Output: ERROR)
- string --> float (Output: ERROR)
- string --> boolean (Output: boolean value)
- string --> complex (Output: ERROR)

5. boolean(if boolean = True) --> int, float, string, complex
   
- boolean --> int (Output: int value)
- boolean --> float (Output: float value)
- boolean --> string (Output: string value)
- boolean --> complex (Output: complex values)

6. boolean(if boolean = False) --> int, float, string, complex
   
- boolean --> int (Output: int value)
- boolean --> float (Output: float value)
- boolean --> string (Output: string value)
- boolean --> complex (Output: complex values)

7. complex --> int, float, string, boolean
   
- complex --> int (Output: int value)
- complex --> float (Output: float value)
- complex --> string (Output: string value)
- complex --> boolean (Output: boolean values)

```python
# int --> float

i = 123              
print (i, type(i))   # int
f = float(i)         # int --> float
print (f, type(f))   # float
```

```python
# int --> string

i = 123
print (i, type(i))    # int 
s = str(i)            # int --> string
print (s, type(s))    # string
```

```python
# int --> boolean (for True)

i = 123
print (i, type(i))    # int 
b = bool(i)           # int --> boolean
print (b, type(b))    # boolean
```

```python
# int --> boolean (for False)

i = 0
print (i, type(i))    # int 
b = bool(i)           # int --> boolean
print (b, type(b))    # boolean
```

```python
# int --> complex

i = 12
print (i, type(i))    # int
c = complex(i)        # int --> complex
print (c, type(c))    # complex
```

```python
# float --> boolean (for True)

f = 100.54
print (f, type(f))    # float
b = bool(f)           # float --> boolean
print (b, type(b))    # boolean
```

```python
# float --> boolean (for False)

f = 0.0
print (f, type(f))    # float 
b = bool(f)           # float --> boolean
print (b, type(b))    # boolean
```

```python
# float --> complex

f = 100.54
print (f, type(f))    # float 
c = complex(f)        # float -- > complex
print (c, type(c))    # complex
```
