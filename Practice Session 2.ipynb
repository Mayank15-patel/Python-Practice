# String Manipulations

```python
s = "Programming"
output = ""

for i in s:
    output += i + "-"

print(output[:-1]) 
```

```python
str17 = "MayankPaTEL"
upper = 0
lower = 0
for i in str17:
    if ord(i) >= 65 and ord(i) <= 90:
        upper+= 1
    elif ord(i) >= 97 and ord(i) <= 122:
        lower+= 1
print ("The Uppercase letter are:", upper)
print ("The Lowercase letter are:", lower)
```

```python
str18 = "mayank15patel2003"
char = 0
digit = 0
for i in str18:
    if ord(i) >= 97 and ord(i) <= 122:
        char+= 1
    elif ord(i) >= 48 and ord(i) <= 57:
        digit+= 1
print ("The Alphabet characters are:", char)
print ("The Digits are:", digit)
```

```python
str21 = "Python is very easy"
new_str21 = ""

for i in str21:
    if i != " ":
        new_str21 += i
        
print(new_str21)
```

```python
s = "Mayank2003patel1507"
output = ""

for i in s:
    if i.isdigit():
        output += "*"
    else:
        output += i

print(output)
```

```python
str23 = "mayankpatelmayurpatel"
vowel = 0
consonent = 0
vowels = "aeiou"
for i in str23:
    if i in vowels:
        vowel = vowel + 1
    else:
        consonent = consonent + 1
print ("The count of vowel is: ",vowel)
print ("The count of Consonents is: ", consonent)
```
