```python
# int --> string

i = 123
print (i, type(i))
s = str(i)
print (s, type(s))
```

```python
# int --> boolean (for True)

i = 123
print (i, type(i))
b = bool(i)
print (b, type(b))
```

```python
# int --> boolean (for False)

i = 0
print (i, type(i))
b = bool(i)
print (b, type(b))
```

```python
# int --> complex

i = 12
print (i, type(i))
c = complex(i)
print (c, type(c))
```
